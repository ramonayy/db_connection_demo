#### Verbindungsaufbau

```python
import mysql.connector
import os
```

```python
# connect to the database
# best practice is the use of environment variables

mydb = mysql.connector.connect(
    host="localhost",
    user="root",
    password="cuckoo5192#",
    database="company"
)

print(mydb)
```

--------------------------------------------------------------------------------------------------------------------------------------------------------------

#### Cursor Objekt

```python
# create a cursor object to interact with the database and execute a query

cursor = mydb.cursor()
cursor.execute("SELECT VName, Nname FROM Angestellter")
```

```python
# read the first row of the result set

row = cursor.fetchone()
print(row)
```

```python
# read the next three rows of the result set

rows = cursor.fetchmany(size=3)
for row in rows:
    print(row)
```

```python
# read all remaining rows of the result set

last_rows = cursor.fetchall()
for row in last_rows:
    print(row)
```

```python
# if the entire result set has been read, the cursor is empty

for x in cursor:
    print(x)
```

```python
# Executing a new query with the same cursor

cursor.execute("SELECT ProjNr, Name, Ort, Abteilung FROM Projekt")
for x in cursor:
    print(x)
```

--------------------------------------------------------------------------------------------------------------------------------------------------------------

#### Result Sets

```python
# generating a second buffered cursor to retrieve data from the Abteilung table

second_cursor = mydb.cursor(buffered=True)
second_cursor.execute("SELECT * FROM Abteilung")

row = second_cursor.fetchone()
print(row)
```

```python
# executing a new query with the old cursor

cursor.execute("SELECT ProjNr, Name, Ort, Abteilung FROM Projekt")
rows = cursor.fetchmany(size=3)
for row in rows:
    print(row)
```

```python
# fetching the remaining rows from the second cursor

for x in second_cursor:
    print(x)
```

```python
# fetching the remaining rows from the fist cursor

rows = cursor.fetchall()
for row in rows:
    print(row)
```

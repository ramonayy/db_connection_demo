# Database Connection Demo

#### Datenbankanbindung mit Python Teil 1 (4-5)
MySQL Connector/Python

```python
import mysql.connector
import os
```

```python
# Connect to the database

mydb = mysql.connector.connect(
    host=os.getenv("DB_HOST"),
    user=os.getenv("DB_USER"),
    password=os.getenv("DB_PASSWORD"),
    database="company"
)

print(mydb)
```

--------------------------------------------------------------------------------------------------------------------------------------------------------------

#### Datenbankanbindung mit Python Teil 1 (9)
Cursor Objekt

```python
# Execute a query to retrieve data from the Angestellter table

cursor = mydb.cursor()
cursor.execute("SELECT VName, Nname FROM Angestellter")

rows = cursor.fetchall()
print(rows)
```

```python
# If the entire Result Set has been read, the cursor will be empty. To read the Result Set again, you need to execute the query again.

for x in cursor:
    print(x)
```

```python
# We only read the first row of the Result Set.

cursor.execute("SELECT VName, Nname FROM Angestellter")

row = cursor.fetchone()			# Ausgabe der nächsten Zeile eines Result Sets
print(row)
```

```python
# We read the next three rows of the Result Set.

selected_rows = cursor.fetchmany(size=3)	# Ausgabe mehrerer Zeilen nach der Anzahl spezifiziert in size
print(selected_rows)
```

```python
# Executing a new query with the same cursor

cursor.execute("SELECT ProjNr, Name, Ort, Abteilung FROM Projekt")
for x in cursor:
    print(x)
```

```python
last_rows = cursor.fetchall()		# Ausgabe aller Zeilen eines Result Sets
print(last_rows)
```

--------------------------------------------------------------------------------------------------------------------------------------------------------------

#### Datenbankanbindung mit Python Teil 1 (10)
Result Sets

```python
# Generating a second buffered cursor to retrieve data from the Abteilung table as a dictionary.

second_cursor = mydb.cursor(dictionary=True, buffered=True)	# Ausgabe der Zeilen als Dictionary
second_cursor.execute("SELECT * FROM Abteilung")

row = second_cursor.fetchone()
print(row)
```

```python
# Executing a new query with the second cursor and fetching its result is possible without reading the entire Result Set of the previous query.

second_cursor.execute("SELECT ProjNr, Name, Ort, Abteilung FROM Projekt")
for row in second_cursor:
    print(row["Name"])
    print(row["Ort"])
```

Abschliessendes

```python
if mydb.is_connected():
    cursor.close()
    second_cursor.close()
    mydb.close()
    print("Database connection closed.")
```

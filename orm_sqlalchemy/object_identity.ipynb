# Objektidentität

```python
class Person:
    def __init__(self, id, name):
        self.id = id
        self.name = name
```

```python
p1 = Person(3, "Nela")
p2 = Person(3, "Nela")

print(p1)
print(p2)
```

```python

p1 == p2
```

```python
p1.__eq__(p2)
```

```python
p1.__hash__() == p2.__hash__()
```

```python
class Person:
    def __init__(self, id, name):
        self.id = id
        self.name = name

    def __eq__(self, other):
        if isinstance(other, Person):
            return self.id == other.id and self.name == other.name
        else:
            return False

    # Achtung: hash() kann nur einen Parameter entgegennehmen, darum müssen self.id und self.name als Tuple übergeben werden!
    def __hash__(self):
        return hash((self.id, self.name))
```

```python
p1 = Person(3, "Nela")
p2 = Person(3, "Nela")
```

```python
p1 == p2
```

```python
p1.__eq__(p2)
```

```python
p1.__hash__() == p2.__hash__()
```
